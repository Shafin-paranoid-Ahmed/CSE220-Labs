```python
# task 1
def playRight(a, b):
  k = 0
  n = 0
  while k < len(b):
    if b[k] == 1:
      n += 1
    k += 1
  while n>0:
    i = len(a)-1
    temp = a[i]
    while i>0:
      a[i] = a[i-1]
      i -= 1
    a[i] = temp
    n -= 1
  return a


print("///  Test 01: Play Right  ///")
sequence = np.array([10, 20, 30, 40, 50, 60])
beats = np.array([1, 0, 0, 1, 0, 1])
returned_value = playRight(sequence, beats)
print(f'Task 1: {returned_value}')  # This should print [40, 50, 60, 10, 20, 30]
```

```python
# task 2

#Lab1 task 2
def discardCards(a,b):
  j = 0
  count = 0
  while count < len(a):
    if a[j] == b:
      k = j
      temp = a[j]
      while k < len(a)-1:
        a[k] = a[k+1]
        k += 1
      count += 1
      a[k] = temp
    else:
      j += 1
  for i in range(len(a)):
    if a[i] == b:
      a[i] = 0

  return a


print("///  Test 02: Discard Cards  ///")
cards = np.array([1,2,3,2,8,2,2,5,7])
returned_value = discardCards(cards, 2)
print(f'Task 2: {returned_value}') # This should print [1, 3, 8, 5, 7, 0, 0, 0, 0]
unittest.output_test(returned_value, np.array([1, 3, 8, 5, 7, 0, 0, 0, 0]))
```

```python
# task 3
def mergeLineup(a, b):
  m = np.zeros(len(a), dtype = int)
  i = 0
  j = len(b)-1
  while i < len(a):
    if a[i] == None:
      a[i] = 0
    if b[j] == None:
      b[j] = 0
    m[i] = a[i] + b[j]
    i += 1
    j -= 1
  return m



print("///  Test 03: Merge Lineup  ///")
pokemon_1 = np.array([12, 3, 25, 1, None])
pokemon_2 = np.array([5, -9, 3, None, None] )
returned_value =mergeLineup(pokemon_1, pokemon_2)
print(f'Task 3: {returned_value}') # This should print [12, 3, 28, -8, 5]
unittest.output_test(returned_value, np.array([12, 3, 28, -8, 5]))

pokemon_1 = np.array([4, 5, -1, None, None])
pokemon_2 = np.array([2, 27, 7, 12, None])
returned_value =mergeLineup(pokemon_1, pokemon_2)
print(f'Task 3: {returned_value}') # This should print [4,17,6,27,2]
unittest.output_test(returned_value, np.array([4,17,6,27,2]))
```

```python
# task 4
def balanceSalami(a):
  count = 0
  flag = None
  while count < len(a):
    sum1 = 0
    sum2 = 0
    if count == 0:
      sum1 += a[0]
    else:
      for i in range(count+1):
        sum1 += a[i]
    if count == len(a)-1:
      sum2 += a[count]
    else:
      for j in range(count+1, len(a)):
        sum2 += a[j]

    if sum1 == sum2:
      flag = True
      break
    else:
      flag = False
    count += 1
  return flag




print("///  Test 04: Balance Your Salami  ///")
salami = np.array([1, 1, 1, 2, 1])
returned_value = balanceSalami(salami)
print(f'Task 4: {returned_value}') # This should print True
unittest.output_test(returned_value, True)

salami = [2, 1, 1, 2, 1]
returned_value = np.array(balanceSalami(salami))
print(f'Task 4: {returned_value}') # This should print False
unittest.output_test(returned_value, False)

salami = [10, 3, 1, 2, 10]
returned_value =  np.array(balanceSalami(salami))
print(f'Task 4: {returned_value}') # This should print True
unittest.output_test(returned_value, True)
```

```python
# task 6
def waveYourFlag(a):
  if a[0]%2 == 0:
    evenC = 0
    oddC = 1
    for i in range(len(a)):
      if a[i]%2 != 0:
        if oddC != i:
          temp = a[i]
          for j in range(i+1, len(a)):
            a[oddC+1] = a[oddC]
          a[oddC] = temp
        oddC += 2
      evenC += 2
  elif a[0]%2 != 0:
      oddC = 0
      evenC = 1
      for i in range(len(a)):
        if a[i]%2 == 0:
          if evenC != i:
            temp = a[i]
            for j in range(i+1, len(a)):
              a[evenC+1] = a[evenC]
            a[evenC] = temp
          evenC += 2
        oddC += 2

  return a



print("///  Test 06: Odd Even Wave  ///")
arr = np.array([2,12,3,8,1,5])
returned_value = waveYourFlag(arr)
print(f'Task 6: {returned_value}') # This should print [2,3,12,1,8,5]
unittest.output_test(returned_value, np.array([2,3,12,1,8,5]))

arr = np.array([45,23,78,84,41])
returned_value = waveYourFlag(arr)
print(f'Task 6: {returned_value}') # This should print [45,78,23,84,41]
unittest.output_test(returned_value, np.array([45,78,23,84,41]))
```

```python
def mean(a):
  i = 0
  sum = 0
  while i < len(a):
    sum += a[i]
    i += 1
  mean = sum/len(a)
  return mean

def standardDeviation(a):
  i = 0
  sum = 0
  m = mean(a)
  while i < len(a):
    sum += ((a[i]-m)**2)
    i += 1
  stanDev = (sum/(len(a)-1))**0.5
  return stanDev

def newArray(a):
  count = 0
  i = 0
  m = mean(a)
  s = standardDeviation(a)
  while i < len(a):
    if (a[i]-m)/s >= 1.5 or (a[i]-m)/s <= -1.5:
      count += 1
    i += 1
  if count == 0:
    return None
  else:
    nA = np.zeros(count, dtype = int)
    j = 0
    k = 0
    while k < len(a):
      temp = (a[k]-m)/s
      if temp >= 1.5 or temp <= -1.5:
        nA[j] = a[k]
        j += 1
      if j == count:
        break
      k += 1
    return nA


a = np.array([10, 8, 13, 9, 14, 25, -5, 20, 7, 7, 4])
print(mean(a))
print(standardDeviation(a))
print(newArray(a))


```
