```python

class Patient:
  def __init__(self, id=None, name=None, age=None, bloodGroup=None, prev= None, next = None):
    self.id = id
    self.name = name
    self.age = age
    self.bg = bloodGroup
    self.next = self if next == None else next
    self.prev = self if prev == None else prev

class WRM:
  patientCount = 0
  dhead = Patient()
  temp = dhead.next
  def RegisterPatient(self, id, name, age, bg):
    new = Patient(id, name, age, bg, WRM.temp, WRM.dhead)
    WRM.temp.next = new
    WRM.dhead.prev = new
    WRM.temp = WRM.temp.next
    WRM.patientCount += 1
    return True

  def ServePatient(self):
    if self.patientCount != 0:
      temp = WRM.dhead.next
      ret = temp.name
      temp.next.prev = WRM.dhead
      WRM.dhead.next = temp.next
      WRM.patientCount -= 1
      return ret
    else:
      return None

  def CancelAll(self):
    WRM.dhead.next, WRM.dhead.prev = WRM.dhead, WRM.dhead
    WRM.temp = WRM.dhead.next
    WRM.patientCount = 0

  def CanDoctorGoHome(self):
    if WRM.patientCount == 0:
      return True
    else:
      return False

  def ShowAllPatient(self):
    if WRM.patientCount != 0:
      temp = WRM.dhead.next
      while temp != WRM.dhead:
        print(temp.name, end = " ")
        temp = temp.next
      print()
    else:
      print("No patient in the wrm")


  def ReverseTheLine(self):
    temp = WRM.dhead.prev
    while temp != WRM.dhead:
      print(temp.name, end = " ")
      temp = temp.prev
    print()




waiting_room_management = WRM()
patient_id = 100
print("**Welcome to Waiting Room Management System**")
while True:
  print()
  print('''==Choose an Option==
1. RegisterPatient()
2. ServePatient()
3. CancelAll()
4. CanDoctorGoHome()
5. ShowAllPatient()
6. exit
======================''')
  woah = int(input("Enter your choice: "))

  if woah == 6:
    print("Thank you for using Waiting Room Management System")
    print("EXITING....")
    break

  elif woah > 6 or woah < 1:
    print("No such option")
    continue
  elif woah == 1:
    print("executing RegisterPatient()")
    id = int(input("Enter ID: "))
    name = input("Enter Name: ")
    age = int(input("Enter Age: "))
    bg = input("Enter Blood Group: ")
    giorno = waiting_room_management.RegisterPatient(id, name, age, bg)
    if giorno == True:
      print("Success registering patient")
    else:
      print("Registering patient unsuccessful")

  elif woah == 2:
    print("executing ServePatient()")
    joestar = waiting_room_management.ServePatient()
    if joestar != None:
      print(f"{joestar} is served")
    else:
      print("No patient to be served")

  elif woah == 3:
    waiting_room_management.CancelAll()
    print("executing CancelAll")
    print("All appointments cancelled")

  elif woah == 4:
    print("executing CanDoctorGoHome()")
    dio = waiting_room_management.CanDoctorGoHome()
    if dio == True:
      print("Yes")
    else:
      print("No")

  elif woah == 5:
    print("executing ShowAllPatient()")
    waiting_room_management.ShowAllPatient()
```
