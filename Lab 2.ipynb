```python
#You must run this cell to print matrix and for the driver code to work
def print_matrix(m):
  row,col = m.shape
  for i in range(row):
    c = 1
    print('|', end='')
    for j in range(col):
      c += 1
      if(len(str(m[i][j])) == 1):
        print(' ',m[i][j], end = '  |')
        c += 6
      else:
        print(' ',m[i][j], end = ' |')
        c += 6
    print()
    print('-'*(c-col))
```

```python
# task 1
def walk_zigzag(floor):
  row, col = floor.shape
  i = 0
  j = 0
  while j < col:
    if j%2 == 0:
      i = 0
      while i < row:
        print(floor[i][j], end= " ")
        i += 2
    elif j%2 != 0:
      if row%2 == 0:
        i = row-1
      elif row%2 != 0:
        i = row-2
      while i > 0:
        print(floor[i][j], end = " ")
        i -= 2
    j += 1
    print()



floor = np.array([[ '3' , '8' , '4' , '6' , '1'],
                  ['7' , '2' , '1' , '9' , '3'],
                  ['9' , '0' , '7' , '5' , '8'],
                  ['2' , '1' , '3' , '4' , '0'],
                  ['1' , '4' , '2' , '8' , '6']]
                )

print_matrix(floor)
print('Walking Sequence:')
walk_zigzag(floor)
#This should print
# 3 9 1
# 1 2
# 4 7 2
# 4 9
# 1 8 6
print('################')
floor = np.array([[ '3' , '8' , '4' , '6' , '1'],
                  ['7' , '2' , '1' , '9' , '3'],
                  ['9' , '0' , '7' , '5' , '8'],
                  ['2' , '1' , '3' , '4' , '0']]
                )

print_matrix(floor)
print('Walking Sequence:')
walk_zigzag(floor)
#This should print
# 3 9
# 1 2
# 4 7
# 4 9
# 1 8
```

```python
# task 2
def create_fence(district, depth):
    row, col = district.shape
    d = depth
    new = np.zeros((row + 2*d,col + 2*d), dtype = int)

    ## the new matrix as a whole 8, cause if it was place the other way
    ## around then if there was a 0 in the district then it will also turn into a 8 ##

    for x in range(row+2*d):
      for y in range(col+2*d):
        if new[x][y] == 0:
          new[x][y] = 8
    i = 0
    while i < row:
      j = 0
      while j < col:
        new[i+d][j+d] = district[i][j]
        j += 1
      i += 1
    return new





depth = 1
district = np.array([[2,3,4], [3,4,6], [2,1,4]])
print_matrix(district)
ans = create_fence(district, depth)
print_matrix(ans)
#This will print
# |  8  |  8  |  8  |  8  |  8  |
# -------------------------------
# |  8  |  2  |  3  |  4  |  8  |
# -------------------------------
# |  8  |  3  |  4  |  6  |  8  |
# -------------------------------
# |  8  |  2  |  1  |  4  |  8  |
# -------------------------------
# |  8  |  8  |  8  |  8  |  8  |
# -------------------------------
print('################')
depth = 2
district = np.array([
                 [2,3,4,1],
                 [3,4,6,5],
                 [2,1,4,7]
                ])
print_matrix(district)
ans = create_fence(district, depth)
print_matrix(ans)
```

```python
# task 3

def strength_difference(clubroom):
  row, col = clubroom.shape
  n = 0
  for x in range(row):
    n += x
  strength_diff = np.zeros(n, dtype = int)

  count1 = 0
  j = 0
  while j < col:
    i = 0
    while i < j:
      strength_diff[count1] = clubroom[i][j] - clubroom[j][i]
      count1 += 1
      i += 1
    j += 1
  return strength_diff

clubroom = np.array([
[1,  2,  9,  7],
[4,  5,  1,  8],
[3,  6,  2,  7],
[2,  8,  6,  3]
])
print_matrix(clubroom)
returned_value = strength_difference(clubroom)
print('Strength Difference Array is : ', returned_value)
unittest.output_test(returned_value, np.array([-2, 6, 5, -5, 0, 1]))
```

```python
# task 4
def atm_triangle(n):
  new = np.zeros((n,n), dtype = int)
  m = 0
  for i in range(0,n):
    m += 1
    for j in range(0,n):
      if j > i:
        continue
      if j == 0 or j == i:
        new[i][j] = m
      else:
        sum = 0
        for k in range(j+1):
          if new[i-1][k] != 0:
            sum += new[i-1][k]
        new[i][j] = sum
  return new

def print_without_none(matrix):
  row, col = matrix.shape
  for i in range(row):
    for j in range(col):
      if matrix[i][j] != 0:
        print(matrix[i][j], end = " ")
    print()

n = 5
returned_value = atm_triangle(n)
print_without_none(returned_value)
#This should print
# 1
# 2  2
# 3  4  3
# 4  7  10  4
# 5  11 21  25  5
```

```python
# task 5

def moving_around(cmds):
  grid = np.full((7, 7), ".")
  grid[3][3] = "-"
  og_pos = np.array((3,3), dtype = int)
  for i in range(len(cmds)):
    if 4 >= cmds[i] >= 1: ##Upper left
      if og_pos[0] - 2 >= 0 and og_pos[1]-2 >= 0:
        og_pos[0] -= 2
        og_pos[1] -= 2
        if cmds[i] == 1 and og_pos[1] - 1 >= 0:
          og_pos[1] -= 1
        elif cmds[i] == 2 and og_pos[1] + 1 < 7:
          og_pos[1] += 1
        elif cmds[i] == 3 and og_pos[0] - 1 >= 0:
          og_pos[0] -= 1
        elif cmds == 4 and og_pos[0] + 1 < 7:
          og_pos[0] += 1
        else:
          og_pos[0] += 2
          og_pos[1] -= 2
        grid[og_pos[0], og_pos[1]] = "*"


    elif 8 >= cmds[i] >= 5: ##Upper right
      if og_pos[0] - 2 >= 0 and og_pos[1] + 2 < 7:
        og_pos[0] -= 2
        og_pos[1] += 2
        if cmds[i] == 5 and og_pos[1] - 1 >= 0:
          og_pos[1] -= 1
        elif cmds[i] == 6 and og_pos[1] + 1 < 7:
          og_pos[1] += 1
        elif cmds[i] == 7 and og_pos[0] - 1 >= 0:
          og_pos[0] -= 1
        elif cmds == 8 and og_pos[0] + 1 < 7:
          og_pos[0] += 1
        else:
          og_pos[0] += 2
          og_pos[1] -= 2
        grid[og_pos[0], og_pos[1]] = "*"


    elif 12>=cmds[i]>=9: ##Lower left
      if og_pos[0]+2 < 7 and og_pos[1]-2 >= 0:
        og_pos[0] += 2
        og_pos[1] -= 2
        if cmds[i] == 9 and og_pos[1] - 1 >= 0:
          og_pos[1] -= 1
        elif cmds[i] == 10 and og_pos[1] + 1 < 7:
          og_pos[1] += 1
        elif cmds[i] == 11 and og_pos[0] + 1 >= 0:
          og_pos[0] -= 1
        elif cmds[i] == 12 and og_pos[0] + 1 < 7:
          og_pos[0] += 1
        else:
          og_pos[0] -= 2
          og_pos[1] += 2
        grid[og_pos[0], og_pos[1]] = "*"

  grid[og_pos[0],og_pos[1]] = "/"

  return grid

cmds = np.array([5,11,2,9])
result = moving_around(cmds)
print_matrix(result)
#This should print
# -------------------------------------------
# |  .  |  /  |  .  |  .  |  .  |  .  |  .  |
# -------------------------------------------
# |  .  |  .  |  .  |  .  |  *  |  .  |  .  |
# -------------------------------------------
# |  .  |  .  |  *  |  .  |  .  |  .  |  .  |
# -------------------------------------------
# |  .  |  .  |  .  |  -  |  .  |  .  |  .  |
# -------------------------------------------
# |  .  |  .  |  .  |  .  |  .  |  .  |  .  |
# -------------------------------------------
# |  .  |  .  |  .  |  .  |  .  |  .  |  .  |
# -------------------------------------------
# |  .  |  .  |  .  |  .  |  .  |  .  |  .  |
# -------------------------------------------
```
